# H1.2-G — Kaggle launcher

Implementation lives in `captioning/`. Enable Internet and GPU; attach datasets. TEST_AFTER_TRAIN controls automatic full test and scoring after training.


```python
import os
import subprocess
import sys
from pathlib import Path

repo = Path('/kaggle/working/Image_Captioning')
if not repo.exists():
    subprocess.run(['git', 'clone', 'https://github.com/Supzxjee/Image_Captioning.git', str(repo)], check=True)
else:
    subprocess.run(['git', '-C', str(repo), 'pull', '--ff-only'], check=True)
os.chdir(repo)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
subprocess.run(['git', 'rev-parse', 'HEAD'], check=True)

# Paste the visual cache directory copied from Kaggle Input, or leave empty for pixels.
VISUAL_CACHE = ""
cache_args = ["--visual-cache", VISUAL_CACHE] if VISUAL_CACHE else []

TEST_AFTER_TRAIN = True  # Full test + metrics after final epoch
test_args = ["--test-after-train"] if TEST_AFTER_TRAIN else []
```

## Train and optional full test
TEST_AFTER_TRAIN=True runs full test and metrics after final epoch. Set False for train only. Captions are saved before scoring.


```python
if VISUAL_CACHE:
    subprocess.run([sys.executable, '-u', 'train_h1_2_gated.py', '--mode', 'verify-cache', '--split', 'val', '--limit', '10'] + cache_args, check=True)
subprocess.run([sys.executable, '-u', 'train_h1_2_gated.py', '--mode', 'train', '--epochs', '10', '--experiment-name', 'h1_2_gated_visualcache' if VISUAL_CACHE else 'h1_2_gated_prompt_to_visual_crossattn'] + cache_args + test_args, check=True)
```
