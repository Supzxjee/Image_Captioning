# H1.2-G — Kaggle launcher

Implementation lives in the `captioning/` Python package. Enable Internet and GPU; attach COCO and prompt cache datasets. Train and evaluation run separately.


```python
import os
import subprocess
import sys
from pathlib import Path

repo = Path('/kaggle/working/Image_Captioning')
if not repo.exists():
    subprocess.run(['git', 'clone', 'https://github.com/Supzxjee/Image_Captioning.git', str(repo)], check=True)
else:
    subprocess.run(['git', '-C', str(repo), 'pull', '--ff-only'], check=True)
os.chdir(repo)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
subprocess.run(['git', 'rev-parse', 'HEAD'], check=True)

# Paste the visual cache directory copied from Kaggle Input, or leave empty for pixels.
VISUAL_CACHE = ""
cache_args = ["--visual-cache", VISUAL_CACHE] if VISUAL_CACHE else []
```

## Train

Only trains and saves checkpoints. No automatic full test. To warm-start H1.2, add `--checkpoint` and use a smaller `--lr`; this resets the optimizer.


```python
if VISUAL_CACHE:
    subprocess.run([sys.executable, '-u', 'train_h1_2_gated.py', '--mode', 'verify-cache', '--split', 'val', '--limit', '10'] + cache_args, check=True)
subprocess.run([sys.executable, '-u', 'train_h1_2_gated.py', '--mode', 'train', '--epochs', '10', '--experiment-name', 'h1_2_gated_visualcache' if VISUAL_CACHE else 'h1_2_gated_prompt_to_visual_crossattn'] + cache_args, check=True)
```
