# Build both complete CLIP prompt caches
Attach ALL full VLM scene shards and dataset_coco.json. Set SCENE_DIR to their Input directory. Smoke shards must be excluded. No visual feature recaching is needed.


```python
import os, sys, subprocess
from pathlib import Path
REPO = Path('/kaggle/working/Image_Captioning')
def run(args):
    print('Running:', ' '.join(map(str, args)), flush=True)
    subprocess.run(list(map(str, args)), check=True)
if not REPO.exists():
    run(['git', 'clone', 'https://github.com/Supzxjee/Image_Captioning.git', REPO])
os.chdir(REPO)
run(['git', 'fetch', 'origin'])
run(['git', 'checkout', 'main'])
run(['git', 'pull', '--ff-only'])
run(['git', 'rev-parse', 'HEAD'])
run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements-vlm.txt'])
```

```python
JSON = '/kaggle/input/datasets/vuthetam/mscoco-2014/dataset_coco.json'
SCENE_DIR = Path('/kaggle/input/your-vlm-scenes')  # CHANGE to your attached scene Dataset.
SCENES = sorted(p for p in SCENE_DIR.rglob('scenes_part_*.jsonl') if 'smoke' not in p.name)
assert SCENES, 'No full scene shards found; update SCENE_DIR.'
run([sys.executable, '-u', 'build_vlm_prompts.py', 'embed',
     '--dataset-json-path', JSON, '--scenes', *SCENES,
     '--output-dir', '/kaggle/working/vlm_prompt_cache'])
# Save the two .pt files and metadata .json files as a Dataset for training.
```
