# Own visual cache — bilinear / CLIP FP32 / FP16 storage

Attach COCO images + JSON; enable Internet/GPU. Use one Save & Run All per part. Outputs are approximately 9.3 GB each. Smoke first in a separate version. No prompt cache needed.


```python
import os, sys, subprocess
from pathlib import Path
repo = Path('/kaggle/working/Image_Captioning')
if not repo.exists():
    subprocess.run(['git', 'clone', 'https://github.com/Supzxjee/Image_Captioning.git', str(repo)], check=True)
os.chdir(repo)
subprocess.run(['git', 'fetch', 'origin'], check=True)
subprocess.run(['git', 'checkout', 'main'], check=True)
subprocess.run(['git', 'pull', '--ff-only'], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', 'requirements.txt'], check=True)
subprocess.run(['git', 'rev-parse', 'HEAD'], check=True)
```

```python
PART = 1  # Run separate versions with PART 1, 2, 3, 4
SMOKE = True  # First test 50 images; switch False in fresh version for full part.
command = [sys.executable, '-u', 'cache_visual_features.py', '--part', str(PART), '--parts', '4',
           '--batch-size', '32', '--output-dir', '/kaggle/working/visual_cache']
if SMOKE:
    command += ['--limit', '50']
subprocess.run(command, check=True)
```
