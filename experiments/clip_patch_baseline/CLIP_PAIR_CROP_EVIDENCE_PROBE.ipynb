```python
"""Pixel-exact crop rectangle and original-image patch coordinates; no model imports."""
import math


def crop_rectangle(subject,obj,width,height):
    assert width>0 and height>0
    for b in (subject,obj):assert len(b)==4 and 0<=b[0]<b[2]<=1 and 0<=b[1]<b[3]<=1
    x0=min(subject[0],obj[0])*width;y0=min(subject[1],obj[1])*height
    x1=max(subject[2],obj[2])*width;y1=max(subject[3],obj[3])*height
    dx=.1*(x1-x0);dy=.1*(y1-y0)
    return [max(0,math.floor(x0-dx)),max(0,math.floor(y0-dy)),min(width,math.ceil(x1+dx)),min(height,math.ceil(y1+dy))]



import torch
from transformers import CLIPModel, CLIPProcessor

device = 'cuda' if torch.cuda.is_available() else 'cpu'
processor = CLIPProcessor.from_pretrained(MODEL, revision=REVISION)
model = CLIPModel.from_pretrained(MODEL, revision=REVISION, use_safetensors=True).to(device).eval()
rows = []
with torch.inference_mode():
    for group in GROUPS:
        first = group['events'][0]
        path = paths[first['filename']]
        with Image.open(path) as original:
            image = original.convert('RGB')
        boxes = [crop_rectangle(e['subject_box'], e['object_box'], *image.size) for e in group['events']]
        views = [image, *(image.crop(box) for box in boxes)]
        # Pad both views to preserve the whole content under CLIP center crop.
        views = [ImageOps.pad(view, (224, 224), method=Image.Resampling.BICUBIC,
                             color=(123, 117, 104)) for view in views]
        inputs = processor(text=group['descriptions'], images=views, padding=True, return_tensors='pt')
        result = model(**inputs.to(device))
        img = torch.nn.functional.normalize(result.image_embeds.float())
        txt = torch.nn.functional.normalize(result.text_embeds.float())
        scores = (img @ txt.T).cpu().tolist()
        if not all(math.isfinite(v) for pair in scores for v in pair):
            raise ValueError(f'Nonfinite similarities for {path.name}')
        margins = {
            mode: [pair[idx] - pair[1-idx] for pair, idx in zip(pairs, group['positive_indices'], strict=True)]
            for mode, pairs in [('full', [scores[0]] * 2), ('crop', scores[1:])]
        }
        rows.append(dict(image_id=first['image_id'], image_sha256=hashlib.sha256(path.read_bytes()).hexdigest(),
                         group_id=group['group_id'], family=group['family'], descriptions=group['descriptions'],
                         positive_indices=group['positive_indices'], crop_xyxy=boxes, similarities=scores, margins=margins))

summary = []
for family in ['all', *sorted({row['family'] for row in rows})]:
    subset = [row for row in rows if family == 'all' or row['family'] == family]
    for mode in ('full', 'crop'):
        per_image = [row['margins'][mode] for row in subset]
        values = [v for pair in per_image for v in pair]
        summary.append(dict(family=family, view=mode, images=len(subset), events=len(values),
                            correct_events=sum(v > 1e-6 for v in values),
                            tied_events=sum(abs(v) <= 1e-6 for v in values),
                            both_correct=sum(all(v > 1e-6 for v in pair) for pair in per_image),
                            mean_margin=sum(values)/len(values)))
payload = dict(model=MODEL, revision=REVISION, device=device, locked_manifest_sha256=LOCKED_SHA,
               splits=['train', 'restval'], summary=summary, records=rows)
OUTPUT.parent.mkdir(parents=True, exist_ok=True)
OUTPUT.write_text(json.dumps(payload, indent=2, allow_nan=False), encoding='utf-8')
print(json.dumps(summary, indent=2))
print('SEND:', OUTPUT)
```
