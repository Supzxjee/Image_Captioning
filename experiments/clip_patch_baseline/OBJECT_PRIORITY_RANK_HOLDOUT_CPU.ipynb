```python

from hashlib import sha256
from itertools import combinations
import json
import math
from pathlib import Path
import numpy as np

def is_holdout(row):
    return sha256(f"priority-rank/1411/{row['coco_id']}".encode()).digest()[0]<64

fit=[r for r in ROWS if not is_holdout(r)]
hold=[r for r in ROWS if is_holdout(r)]
assert not ({r['coco_id'] for r in fit} & {r['coco_id'] for r in hold})
classes={name:i for i,name in enumerate(sorted({s['label'] for r in fit for s in r['slots']}))}
base=np.array([[math.log(s['area']),s['confidence']] for r in fit for s in r['slots']])
center=base.mean(axis=0)
scale=np.maximum(base.std(axis=0),1e-8)

def features(slot, with_class):
    numeric=(np.array([math.log(slot['area']),slot['confidence']])-center)/scale
    if not with_class:
        return numeric
    label=np.zeros(len(classes))
    i=classes.get(slot['label'])
    if i is not None:
        label[i]=1
    return np.concatenate((numeric,label))

def train(with_class):
    x,y=[],[]
    for row in fit:
        pairs=[(a,b) for a,b in combinations(row['slots'],2) if a['mentions']!=b['mentions']]
        weight=1/math.sqrt(len(pairs))
        for a,b in pairs:
            x.append(weight*(features(a,with_class)-features(b,with_class)))
            y.append(weight*(a['mentions']-b['mentions'])/5)
    x=np.array(x)
    beta=np.linalg.solve(x.T@x+np.eye(x.shape[1]),x.T@np.array(y))
    assert np.isfinite(beta).all()
    return beta

models={flag:train(flag) for flag in (False,True)}
methods={
    'area':lambda s:s['area'],
    'confidence':lambda s:s['confidence'],
    'ridge_geometry_confidence':lambda s:float(features(s,False)@models[False]),
    'ridge_with_class':lambda s:float(features(s,True)@models[True]),
}
def dcg(slots,score):
    ordered=sorted(slots,key=lambda s:(-score(s),s['roi']))[:3]
    return sum(s['mentions']/math.log2(i+2) for i,s in enumerate(ordered))

per_image=[]
for row in hold:
    slots=row['slots']
    denom=dcg(slots,lambda s:s['mentions'])
    scores={k:dcg(slots,f)/denom for k,f in methods.items()}
    per_image.append(dict(coco_id=row['coco_id'],ndcg_at_3=scores))

sample=np.random.default_rng(1411).integers(0,len(hold),size=(2000,len(hold)))
result={}
for name in methods:
    values=np.array([r['ndcg_at_3'][name] for r in per_image])
    diff=values-np.array([r['ndcg_at_3']['area'] for r in per_image])
    lo,hi=np.quantile(diff[sample].mean(axis=1),[.025,.975])
    result[name]=dict(ndcg_at_3=float(values.mean()),delta_vs_area=float(diff.mean()),ci95=[float(lo),float(hi)])
payload=dict(fit_images=len(fit),held_out_images=len(hold),results=result,
             meets_gate=result['ridge_with_class']['ci95'][0]>0,per_image=per_image)
out=Path('/kaggle/working/object_priority_rank_holdout.json')
out.write_text(json.dumps(payload,indent=2,allow_nan=False),encoding='utf-8')
print(json.dumps(result,indent=2))
print('SEND:',out)
```
