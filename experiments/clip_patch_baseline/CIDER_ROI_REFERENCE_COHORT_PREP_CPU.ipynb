```python
ARCHIVED = {2587: {'targets': ['donut'], 'foils': ['banana'], 'caption': 'a banana and a banana in a plastic bag', 'image_sha256': 'bc37072a4f32bc99d6b02a684dd9bc69bd63567382315e5f497c9bb4a44c7208', 'reference_mentions': {'donut': 3}}, 3067: {'targets': ['sandwich'], 'foils': ['cat'], 'caption': 'a cat standing on a desk next to a phone', 'image_sha256': '6696917b3e8ba513c18c5b9b9fe58d3a9c1d2ee6194de29d2c62eb1a53736cb4', 'reference_mentions': {'sandwich': 5}}, 10693: {'targets': ['car'], 'foils': ['umbrella'], 'caption': 'a woman holding an umbrella standing in the rain', 'image_sha256': '167de934a475457071c79b06eecc1fc3570ac3e2e3562ae5fc9ffc7abb0bc966', 'reference_mentions': {'car': 2}}, 13296: {'targets': ['bicycle'], 'foils': ['bus'], 'caption': 'a man riding a bike down a street next to a bus', 'image_sha256': '136b05cddc53aac9454f732ac1ebde519944a9ec47ee12f360ad796e84e6d2f4', 'reference_mentions': {'bicycle': 5}}, 14821: {'targets': ['chair'], 'foils': ['backpack'], 'caption': 'two pictures of a backpack and a bag of food', 'image_sha256': '09cc845bbf0c7b45b0bdacbdca1c8817ab0b7ed576e645d735c0b98213847418', 'reference_mentions': {'chair': 2}}, 14941: {'targets': ['chair'], 'foils': ['teddy bear'], 'caption': 'a child sitting on a couch with a teddy bear', 'image_sha256': '3a93ffeb40491cf7292916a0703219a513b63867b28de89b7105464a10b5c01d', 'reference_mentions': {'chair': 4}}, 17244: {'targets': ['bicycle'], 'foils': ['bus'], 'caption': 'a man riding a bike next to a bus', 'image_sha256': '2434b256bb27a67bc9dacce22c4793745dd11d57d61676dd367cbecbc5a14688', 'reference_mentions': {'bicycle': 5}}, 18783: {'targets': ['bicycle'], 'foils': ['dog'], 'caption': 'a man riding a bike with a dog', 'image_sha256': '45f81a9496a5d6e1e750c61ef6dd56b114be96ccdae0b2294665ded103102842', 'reference_mentions': {'bicycle': 5}}, 21452: {'targets': ['car'], 'foils': ['motorcycle'], 'caption': 'a group of motorcycles parked on top of a green field', 'image_sha256': 'bad6142e7fa76d86e076aea0499b04dadb1db83dbe662769181934f3aa3eaca7', 'reference_mentions': {'car': 4}}, 22420: {'targets': ['motorcycle'], 'foils': ['book'], 'caption': 'a couple of books sitting on top of a book shelf', 'image_sha256': 'c07333f38a6fb52a3816e06cc21be3ad0251547215567a1398e726dc46b2c53d', 'reference_mentions': {'motorcycle': 2}}, 25433: {'targets': ['car'], 'foils': ['bus'], 'caption': 'a white bus sitting in front of a building', 'image_sha256': '727a44b8a46c6fec14366e37e96b931c6b5decae0f118ded8864b7f0644e2806', 'reference_mentions': {'car': 2}}, 26363: {'targets': ['umbrella'], 'foils': ['elephant'], 'caption': 'a group of people riding on the backs of elephants', 'image_sha256': 'b7bfb6ccb790426a28954af238dcb5bb1460922c27db8165a0773c4f4dd289a4', 'reference_mentions': {'umbrella': 2}}, 28864: {'targets': ['banana'], 'foils': ['bowl'], 'caption': 'a white bowl filled with apples and berries', 'image_sha256': 'be45c8d4e5a797b1c8bbc811ca12fcbcaa62cfcfaf1062fa87deca560bb5621b', 'reference_mentions': {'banana': 2}}, 33697: {'targets': ['backpack'], 'foils': ['umbrella'], 'caption': 'a green umbrella sitting on top of a sandy beach', 'image_sha256': '0a492439e1207894270ccb9cbc2e2d08708646fcebadaa007df33be8585fef0a', 'reference_mentions': {'backpack': 2}}, 33924: {'targets': ['bed'], 'foils': ['cat'], 'caption': 'a person holding a toothbrush next to a cat', 'image_sha256': '7898f8fe131e61540d2d7d39706e1db0d59376c0cdf08fd16a54a44c5b5b7ce8', 'reference_mentions': {'bed': 3}}, 54485: {'targets': ['car'], 'foils': ['bus'], 'caption': 'a city bus parked on the side of a street', 'image_sha256': '83554821b3edec2479eb9a9218c7148dd4c265b43df7337fb7e30f7c6386b1a4', 'reference_mentions': {'car': 2}}, 71283: {'targets': ['chair'], 'foils': ['tv', 'couch'], 'caption': 'a living room with a couch and a television', 'image_sha256': 'ba2e4b6027cb91855fe1fcdab9e0b967434643e56af240a4e002ad5091c43ab3', 'reference_mentions': {'chair': 2}}, 75636: {'targets': ['bowl'], 'foils': ['pizza'], 'caption': 'a pizza and a salad on a table', 'image_sha256': 'ac75073a9dd22bf0b7289a09b8570ad7a0177569a7fde1b9fd7c629d8ac0fda6', 'reference_mentions': {'bowl': 2}}, 88183: {'targets': ['chair'], 'foils': ['couch'], 'caption': 'two men sitting on a couch in a living room', 'image_sha256': '7bf56de32743d4ddf73288f5a044af11775a4ab2e2b256890d42f61915f4a287', 'reference_mentions': {'chair': 2}}, 88244: {'targets': ['tv'], 'foils': ['laptop'], 'caption': 'a room with a desk and a laptop', 'image_sha256': '6b0013bf26d9f7fe4d53a9eec10e8b141e450f154d63e9af734ee60482fa27f5', 'reference_mentions': {'tv': 2}}, 89541: {'targets': ['banana'], 'foils': ['bowl'], 'caption': 'a yellow bowl filled with fruit and a fruit', 'image_sha256': 'c8fe8e9a6a586f2b8391555b32e162fcbf7efffb838e89ebcb0c27795be350c4', 'reference_mentions': {'banana': 3}}, 94926: {'targets': ['backpack'], 'foils': ['bear'], 'caption': 'a woman in a purple jacket and a white polar bear', 'image_sha256': 'c2958c763f19f19a68445074c79272b266eb83113ddd93c6093ea36a6bf7788d', 'reference_mentions': {'backpack': 2}}, 95692: {'targets': ['umbrella'], 'foils': ['skateboard'], 'caption': 'a man riding a skateboard next to a body of water', 'image_sha256': 'a7408f78d5d17d3e68264ede636469f6e01522e2fd69619cee03edc7c072902e', 'reference_mentions': {'umbrella': 3}}, 97337: {'targets': ['couch'], 'foils': ['tv'], 'caption': 'a living room with a television and a glass table', 'image_sha256': '0bec64b322b803e72a1b1ec23b4daba6e61f8f55ef3a33164d87ae3694ff5d97', 'reference_mentions': {'couch': 2}}, 98054: {'targets': ['cup'], 'foils': ['fork'], 'caption': 'a white plate topped with carrots and a fork', 'image_sha256': 'cbd5bafe8f14ab4116518a502fd8351f87615056df7c7d76a6d15e80ec04c92a', 'reference_mentions': {'cup': 2}}, 107226: {'targets': ['bicycle'], 'foils': ['dog'], 'caption': 'a man on a bike being pulled by three dogs', 'image_sha256': '285a3dfd666d35fb60eb63b9dd46b7b49aae4fcafe194f51d7e4b446abdab725', 'reference_mentions': {'bicycle': 3}}, 112608: {'targets': ['boat'], 'foils': ['horse'], 'caption': 'a group of people riding horses on a beach', 'image_sha256': '49caee7a7d0c067918530e5ab56ed2bd7f3ed62a76bca5ea74fe861fe8519e31', 'reference_mentions': {'boat': 2}}, 117530: {'targets': ['laptop'], 'foils': ['chair'], 'caption': 'a living room with a keyboard and a chair', 'image_sha256': 'bd113edb672757c8ba18b92ef7147b334967f87059a43ff6a2caaf15a671b95e', 'reference_mentions': {'laptop': 2}}, 123842: {'targets': ['cow'], 'foils': ['tv'], 'caption': 'two animals walking in front of a television', 'image_sha256': 'f40645e2f9d38dc5bba48c9eae54ac31838beff988851b942a8ad63685fc6952', 'reference_mentions': {'cow': 3}}, 124210: {'targets': ['bench'], 'foils': ['suitcase'], 'caption': 'a green chair and a blue suitcase on a sidewalk', 'image_sha256': '928b7270a3c3f7b38a1cce9184d12fbbb3e0345b2415363433fedf155252bfc5', 'reference_mentions': {'bench': 4}}, 127337: {'targets': ['bed'], 'foils': ['couch'], 'caption': 'a living room with a couch and a lamp', 'image_sha256': 'f4be0f3c77f6589bf3ef9bb1bc4e7eab27509132ba5fb31ebe081562e342b7b2', 'reference_mentions': {'bed': 2}}, 129566: {'targets': ['chair'], 'foils': ['clock'], 'caption': 'a living room filled with furniture and a clock', 'image_sha256': 'a3a4166502af02355eaa19922363e6a4637c0a80e1e74dba5faf58b19ba31113', 'reference_mentions': {'chair': 4}}, 129800: {'targets': ['car'], 'foils': ['truck'], 'caption': 'a blue truck parked in a parking lot', 'image_sha256': 'dec8fa01a21618e093761fd08df4223198771cea17a4ebfd00f33d8daec3edec', 'reference_mentions': {'car': 3}}, 132059: {'targets': ['spoon'], 'foils': ['cup', 'donut'], 'caption': 'a plate with a donut and a cup of coffee', 'image_sha256': 'b33824debb71b7f0d59c2013494db31b9a3024648fd1b68ba5f36f5e90e9bbd5', 'reference_mentions': {'spoon': 2}}, 140203: {'targets': ['truck'], 'foils': ['bus'], 'caption': 'a group of people standing next to a white bus', 'image_sha256': '5c3a04dada4c36aba47e5c65a25da677451e66c4a7344ebb602bb4fe4cda55d1', 'reference_mentions': {'truck': 2}}, 143653: {'targets': ['horse'], 'foils': ['train'], 'caption': 'a train traveling through a lush green forest', 'image_sha256': 'e211a16ccb0136fa97a72deb1c8615bc84e137e1906bfcabd7201f15c93b5ad2', 'reference_mentions': {'horse': 2}}, 145061: {'targets': ['car'], 'foils': ['bus'], 'caption': 'a bus that is sitting in the street', 'image_sha256': '46b007a447ea0b56bbc09fc0a5edadef7b8dd5b5de92e15bd9ec0d66a13a808a', 'reference_mentions': {'car': 4}}, 147675: {'targets': ['bird'], 'foils': ['boat'], 'caption': 'a group of seagulls flying on a blue boat', 'image_sha256': '77914b9aaa96751a4c308140a9ae19b2e6161d4872b00e449ce0243569b48fd5', 'reference_mentions': {'bird': 4}}, 155145: {'targets': ['umbrella'], 'foils': ['boat'], 'caption': 'a small boat in the water with people in it', 'image_sha256': '728cac372405c5a25ea4f77f2fa29c19e6d2a3b50177d1b7b1f91498773a4591', 'reference_mentions': {'umbrella': 2}}, 172079: {'targets': ['cup'], 'foils': ['vase'], 'caption': 'a group of vases sitting on a table', 'image_sha256': '6d0caf1dd23cb254f956d21c62e3b73dfa8f0813b0714f69b9d39a5f6e30a92d', 'reference_mentions': {'cup': 2}}, 173434: {'targets': ['couch'], 'foils': ['tv'], 'caption': 'a living room with a fireplace and a television', 'image_sha256': 'afced0d2e1793c5c52a62bb2e8b5064cdc52421d3d82b4cc75c8b426dafb2294', 'reference_mentions': {'couch': 2}}, 173464: {'targets': ['bicycle'], 'foils': ['couch'], 'caption': 'a living room with a couch and a table', 'image_sha256': '12d758447826de154cd54fa4c09d2f207793038935be6e6d9a631a42c2c57559', 'reference_mentions': {'bicycle': 2}}, 173545: {'targets': ['umbrella'], 'foils': ['elephant'], 'caption': 'two elephants with people riding on their backs', 'image_sha256': '0ce20d2cc9976a592483cbd8d9bcb0ffde3fe83b0506b9fc7179a92a51c1b47c', 'reference_mentions': {'umbrella': 2}}, 182805: {'targets': ['umbrella'], 'foils': ['dog'], 'caption': 'a woman sitting on a beach next to two dogs', 'image_sha256': 'e502d762b1a264156f1e6349702d742d6cdddd40a2b54a3390acbd8e5059d1dd', 'reference_mentions': {'umbrella': 3}}, 183676: {'targets': ['bicycle'], 'foils': ['surfboard'], 'caption': 'a woman sitting on a beach next to a surfboard', 'image_sha256': 'b766766a38f000e69a859d11c6b5629a67662c9b2bc2eca8c6bfb4f6f0899807', 'reference_mentions': {'bicycle': 5}}, 184659: {'targets': ['motorcycle'], 'foils': ['cake'], 'caption': 'a birthday cake with a train on a table', 'image_sha256': '42aa59230f7eb2bde80bf4f4304d77d1b4933fec565d044d7430a2f9e3c56034', 'reference_mentions': {'motorcycle': 3}}, 184945: {'targets': ['bicycle'], 'foils': ['motorcycle'], 'caption': 'a group of motorcycles parked next to a street', 'image_sha256': '120a515bf507e4e00e9512feb5a80cec9faa6a7c63cab7cccf3edcdf31cd081e', 'reference_mentions': {'bicycle': 5}}, 189452: {'targets': ['bicycle'], 'foils': ['banana'], 'caption': 'a group of people standing around many bunches of bananas', 'image_sha256': '82829cee96c7c0e003e0dcb7e57d27f06071ca43ca9fad8d1c2c051bd7c3fbcf', 'reference_mentions': {'bicycle': 3}}, 190760: {'targets': ['donut'], 'foils': ['cat'], 'caption': 'a person holding a doughnut with a cat', 'image_sha256': 'a2e8bb31ff3cbe07092d713af2e4f5378a637e08a348bcb3b1a4420990fb9506', 'reference_mentions': {'donut': 3}}, 197997: {'targets': ['bicycle'], 'foils': ['bus'], 'caption': 'an old blue bus parked on the side of a road', 'image_sha256': '156be9f7cb2f8c3d437775925608aa12b4f47075f6be9b97fe3e8ddc0506327d', 'reference_mentions': {'bicycle': 2}}, 200231: {'targets': ['bench'], 'foils': ['teddy bear'], 'caption': 'a group of teddy bears sitting next to a basket of flowers', 'image_sha256': '0f483acd6c7a4214f2364c18dbe84f0359465cbdbc7f2138651bc097afcd2c4d', 'reference_mentions': {'bench': 4}}, 202225: {'targets': ['chair'], 'foils': ['tv'], 'caption': 'a living room filled with furniture and a flat screen tv', 'image_sha256': 'c2202922f34e8906ee6309bfad9a533b8e6b50d60ddec4bcb5aa92738c2adb21', 'reference_mentions': {'chair': 2}}, 204256: {'targets': ['book'], 'foils': ['handbag', 'couch'], 'caption': 'a purse sitting on a couch next to a cell phone', 'image_sha256': '3b0beac1025ee3d7d345077b5d8e695d2420da36c7c37655e41e8e1bb19c9268', 'reference_mentions': {'book': 2}}, 210920: {'targets': ['suitcase'], 'foils': ['laptop'], 'caption': 'a laptop computer sitting on the floor next to a black chair', 'image_sha256': 'ae591945f70667be456286b8cd58ae98a9cefbf20eeae5ad4be1231a12306d09', 'reference_mentions': {'suitcase': 2}}, 214363: {'targets': ['bicycle'], 'foils': ['umbrella'], 'caption': 'a group of people walking down a street holding an umbrella', 'image_sha256': 'a9d3f2410d951d281600d383e85d8563b1f49d891b209a028940232a46de0512', 'reference_mentions': {'bicycle': 4}}, 220732: {'targets': ['truck'], 'foils': ['bus'], 'caption': 'a bus parked on the side of the road', 'image_sha256': '81b9254c750f52ebc8fcc5fa421ab8a3ac430db19f541bfb7541adada76a6a30', 'reference_mentions': {'truck': 3}}, 229997: {'targets': ['bird'], 'foils': ['bear'], 'caption': 'a large black bear sitting in the grass', 'image_sha256': '9c62105390761ac4adf9cb96e80652b55f64866c897cb5d6b3de405fe92db5f8', 'reference_mentions': {'bird': 2}}, 230056: {'targets': ['boat'], 'foils': ['bird'], 'caption': 'a group of three birds walking across a beach', 'image_sha256': '05c204d728cd759de89029146ed35875fdb08486ec5338a4b02d140fe165bb5d', 'reference_mentions': {'boat': 2}}, 232993: {'targets': ['bowl'], 'foils': ['pizza'], 'caption': 'a young boy sitting at a table eating pizza', 'image_sha256': 'fe660fe3563fe4fedeb9cda1ce23d7ea821a0a135cf591f0ee05154cb9e55b73', 'reference_mentions': {'bowl': 2}}, 233311: {'targets': ['bowl'], 'foils': ['teddy bear'], 'caption': 'a brown teddy bear sitting on a table next to a pile of oranges', 'image_sha256': '675539b77e294a60f9773251461528f9b62373a9fbd4b25bcc5ca86a61b8121a', 'reference_mentions': {'bowl': 4}}, 237864: {'targets': ['dog'], 'foils': ['elephant'], 'caption': 'a large elephant walking next to a baby elephant', 'image_sha256': '5d375f4edd3876c858f2738b887b7026834cae709b9983880a4ed7f2d0375d4f', 'reference_mentions': {'dog': 4}}, 257941: {'targets': ['fork'], 'foils': ['cup', 'cake'], 'caption': 'a piece of cake and a cup of coffee', 'image_sha256': 'fe174ecbddd74b3cf2d59500118ec4feb610db0f2d7de8cbab8e6f3ed5ac88c6', 'reference_mentions': {'fork': 4}}, 261068: {'targets': ['sheep'], 'foils': ['dog'], 'caption': 'a woman and a dog playing with a dog', 'image_sha256': '501cdb698c42159da8a7d723f2364c5987e49d7f50d3c82a159f566cebf34a42', 'reference_mentions': {'sheep': 2}}, 261707: {'targets': ['chair'], 'foils': ['laptop'], 'caption': 'a laptop computer sitting on top of a table', 'image_sha256': '95a5a74b7e1c88974d443a0d759efc14e25fb491469f064e4790d59f2321df30', 'reference_mentions': {'chair': 3}}, 263613: {'targets': ['cow'], 'foils': ['horse'], 'caption': 'a woman sitting in a doorway next to a horse', 'image_sha256': '5e33384b3d14bce8733aea7f065879e14df4fbfe9a58e532460ddf6be0302c9c', 'reference_mentions': {'cow': 3}}, 266117: {'targets': ['chair'], 'foils': ['couch'], 'caption': 'a living room filled with a black leather couch', 'image_sha256': '071392271adba7031fe0e30e433f170a2939cf66b2dd69057be9c1bd4e1b70f5', 'reference_mentions': {'chair': 2}}, 266250: {'targets': ['tv'], 'foils': ['cat'], 'caption': 'a black cat sitting in front of a person', 'image_sha256': 'ead6f62be986662a80f649b88ec810ced80ae4e26149e114e4956d29f883dc1b', 'reference_mentions': {'tv': 4}}, 274083: {'targets': ['car'], 'foils': ['bus'], 'caption': 'a group of people standing in front of a double decker bus', 'image_sha256': 'e3756f2e81fcc442bab2addcd82ddf27b7a40834bda3217bfef04e5e143df70e', 'reference_mentions': {'car': 3}}, 279145: {'targets': ['cat'], 'foils': ['bench'], 'caption': 'a bench sitting next to a bunch of potted plants', 'image_sha256': '7263822155604dc51b682d4d3e1c35186e42d7ea21195ba9b221b0c057556998', 'reference_mentions': {'cat': 3}}, 280370: {'targets': ['cow'], 'foils': ['horse'], 'caption': 'two men riding on the backs of brown horses', 'image_sha256': 'd4ba4c9881be4fd5a13316422554ec328f90147c7d5a8a618c1093312fd64592', 'reference_mentions': {'cow': 4}}, 283168: {'targets': ['fork'], 'foils': ['sandwich'], 'caption': 'a plastic container filled with a sandwich next to a plastic container of chocolate sauce', 'image_sha256': '23dba871248301bb3794334349c3ea089ff001fbd36e3dbae1cb8f0e154f7053', 'reference_mentions': {'fork': 2}}, 287560: {'targets': ['chair'], 'foils': ['toilet'], 'caption': 'two pictures of a bathroom with a toilet and a window', 'image_sha256': '3a2ac858abd53057631863b5b3b2720959f9402c488e8b0855b9ac6e12704c25', 'reference_mentions': {'chair': 2}}, 288481: {'targets': ['chair'], 'foils': ['vase'], 'caption': 'a table topped with a vase of flowers', 'image_sha256': '788656d470482eb974e96f55bff7a77d58973c4c532dea403866ecd2d69eb643', 'reference_mentions': {'chair': 2}}, 292616: {'targets': ['chair'], 'foils': ['bird'], 'caption': 'a group of birds sitting on a metal bench', 'image_sha256': '73654e118b978612307d256dcbce52367af58b79878ca4ac074b9f62ec02bf0b', 'reference_mentions': {'chair': 5}}, 295194: {'targets': ['cow'], 'foils': ['horse'], 'caption': 'two people riding on the backs of brown horses', 'image_sha256': '360e0c8b4e19b2e72b0b027704ca07af03b43a56ba070e006e5c14f3a821660e', 'reference_mentions': {'cow': 3}}, 295693: {'targets': ['cat'], 'foils': ['couch'], 'caption': 'a woman sitting on a couch in a living room', 'image_sha256': '670f3c722fcb9cd7760bcd7432e9b945b701f364e73adb9c39b97ef856bcf19b', 'reference_mentions': {'cat': 2}}, 301282: {'targets': ['spoon'], 'foils': ['pizza'], 'caption': 'a young girl sitting at a table eating a pizza', 'image_sha256': '7370948af46f81e1a90075585d260d386f12904a72da59c318e88e7305b1ee53', 'reference_mentions': {'spoon': 2}}, 302219: {'targets': ['backpack'], 'foils': ['handbag'], 'caption': 'a back of a car filled with bags and bags', 'image_sha256': '7d9094bacdf89dfbbb0bf565dfd3cd1a51234ffb3078081b4e8f75653d41f8da', 'reference_mentions': {'backpack': 2}}, 314515: {'targets': ['umbrella'], 'foils': ['car', 'bicycle'], 'caption': 'a car with a bunch of bicycles on top of it', 'image_sha256': '5caca6b87d667a033216038267e93a877cbf2aaacb7d00b915095c1c05fd1c9b', 'reference_mentions': {'umbrella': 3}}, 319687: {'targets': ['bench'], 'foils': ['vase'], 'caption': 'a couple of vases with a flower in them', 'image_sha256': 'd148b345e465a2c5d767acab0e50b551c8de9d92609319102dd81324806369e3', 'reference_mentions': {'bench': 4}}, 333556: {'targets': ['cup'], 'foils': ['clock'], 'caption': 'a large clock on a wall next to a counter', 'image_sha256': '686fa3dbdea8a805eea0a97ae8650d61bf6b1643ee3967ac2458bd8280d22629', 'reference_mentions': {'cup': 2}}, 335119: {'targets': ['car'], 'foils': ['bus'], 'caption': 'a bus is parked next to a fence', 'image_sha256': 'fa7f49eb9a0845a786cee8a6cc2e31c3c2154bd8b0a6c84f7e5b7869ff58e8c5', 'reference_mentions': {'car': 2}}, 335670: {'targets': ['car'], 'foils': ['motorcycle', 'truck'], 'caption': 'a truck driving down a street next to a motorcycle', 'image_sha256': 'd5daa0bc2ae1b54f3d498938b952cc2eb59810630ce89a5443d9648eb3e4d1ce', 'reference_mentions': {'car': 2}}, 338986: {'targets': ['car'], 'foils': ['train'], 'caption': 'a blue and yellow train passing a train station', 'image_sha256': '62fee3f216c3cb2af25eac42d4b1b48acf30652ea1e8dd154aa4dcca480b852f', 'reference_mentions': {'car': 2}}, 343198: {'targets': ['umbrella'], 'foils': ['suitcase'], 'caption': 'a piece of luggage sitting next to a red suitcase', 'image_sha256': 'e70d4976c7cca9f4010288e00cc30bd2e21f2f8771af370f76f0826aabf47532', 'reference_mentions': {'umbrella': 2}}, 345848: {'targets': ['cow'], 'foils': ['motorcycle'], 'caption': 'a group of people riding motorcycles down a street', 'image_sha256': '9e19ae0b89f6a1715255849641333e16a72cc518fff062bed10abd616f07518b', 'reference_mentions': {'cow': 2}}, 350000: {'targets': ['chair'], 'foils': ['umbrella'], 'caption': 'a group of umbrellas sitting on top of a beach', 'image_sha256': '4a8cda159f2c304395cfb59700f8acf513861bf578a81c28302ce2dc13331fb7', 'reference_mentions': {'chair': 2}}, 357229: {'targets': ['fork'], 'foils': ['pizza'], 'caption': 'a pizza sitting on top of a pan on a table', 'image_sha256': '40147bf00e379d205fcf44729fb437703bfe123454585410d636be22ccdb0928', 'reference_mentions': {'fork': 3}}, 360705: {'targets': ['bird'], 'foils': ['cow'], 'caption': 'a field full of cows grazing on a lush green field', 'image_sha256': '091e5693279d5b76d74225d14f461fed73b92fb49bc69422942a3d46d5c39c35', 'reference_mentions': {'bird': 2}}, 360899: {'targets': ['bus'], 'foils': ['motorcycle'], 'caption': 'a boy standing next to a row of motorcycles', 'image_sha256': 'b4694d7e040ba877f256cea69d8168160e10a1f972936fd5b5bab3a334e0792a', 'reference_mentions': {'bus': 2}}, 361124: {'targets': ['truck'], 'foils': ['umbrella'], 'caption': 'a man walking in the rain with an umbrella', 'image_sha256': 'c225f63b51ada7b1d0d813c5131d64194a907e37e41f852b36cd744615b5b380', 'reference_mentions': {'truck': 4}}, 362289: {'targets': ['donut'], 'foils': ['cat'], 'caption': 'a cat sitting on a chair and a person holding a piece of food', 'image_sha256': 'c0e579a13e04e1d1bcadc88e2a8588b79476d2a15ed4bd103f44e26aa86f00a0', 'reference_mentions': {'donut': 3}}, 363461: {'targets': ['chair'], 'foils': ['laptop'], 'caption': 'a computer desk topped with a laptop computer', 'image_sha256': '0d41b0b09df6077778df7c39231ce37ecfc7c30273c55c70926fab0a88515eb8', 'reference_mentions': {'chair': 3}}, 369185: {'targets': ['chair'], 'foils': ['couch'], 'caption': 'a living room with a fireplace and a couch', 'image_sha256': '379b94058c5921c03aee20b4bf7bde31a4c440e9ea1d2b72e806c03371e339ae', 'reference_mentions': {'chair': 2}}, 370478: {'targets': ['suitcase'], 'foils': ['laptop'], 'caption': 'a laptop sitting on the floor next to a dresser', 'image_sha256': 'f5ed1c21f176378a77e5e3dcd071908850d193cc200232790071fea5a1234aaf', 'reference_mentions': {'suitcase': 2}}, 370596: {'targets': ['bicycle'], 'foils': ['motorcycle'], 'caption': 'a motorcycle parked on the side of a street', 'image_sha256': '4cd2bf55be55fcf681ac7c1871f6d7d86d2b0e13b249b5312702b9ee269306d3', 'reference_mentions': {'bicycle': 5}}, 373748: {'targets': ['refrigerator'], 'foils': ['tv'], 'caption': 'a kitchen with a television on the wall', 'image_sha256': '79aa8e3f00ccaf0ae0ee6ed4e1c15539c6ffcd1a26fb1677e1a61d5405aecf6c', 'reference_mentions': {'refrigerator': 2}}, 376247: {'targets': ['cow'], 'foils': ['horse'], 'caption': 'a black and white photo of two children and a horse', 'image_sha256': '527a5742a8615bbbc743fd1df3eb1945a937b38042fb7eed9d4c52316f3ffb57', 'reference_mentions': {'cow': 4}}, 377999: {'targets': ['bicycle'], 'foils': ['boat'], 'caption': 'a white sail boat sitting on top of a body of water', 'image_sha256': '997f4df838f9905c2f10aa876c65f78d62fddba74e6801f86ca5cb9a6bba8bac', 'reference_mentions': {'bicycle': 3}}, 381037: {'targets': ['bottle'], 'foils': ['cat', 'refrigerator'], 'caption': 'a cat laying on a counter next to a refrigerator', 'image_sha256': '74e6968f26937cd7ab76419329d8a36f544799e13d57893af5675ce2d74e58a8', 'reference_mentions': {'bottle': 4}}, 382443: {'targets': ['bottle'], 'foils': ['dog'], 'caption': 'a black and white dog sitting next to a building', 'image_sha256': 'a126981e769f11f510de51d3e28275810ac81d7222dbeaf9e440d50c833b51f7', 'reference_mentions': {'bottle': 2}}, 391225: {'targets': ['boat'], 'foils': ['surfboard'], 'caption': 'a man riding a wave on top of a surfboard', 'image_sha256': 'c684d07b753a10fdb25dfced40fcbcef60ac842958e7428d6de8f550f7e7d3bf', 'reference_mentions': {'boat': 2}}, 391415: {'targets': ['truck'], 'foils': ['car'], 'caption': 'a red car towing a red car down a street', 'image_sha256': '811161f176aa117047d8442132c5007dcc3c47b79026935bbe74fc0829018756', 'reference_mentions': {'truck': 5}}, 392108: {'targets': ['dog'], 'foils': ['surfboard'], 'caption': 'a woman holding a surfboard in the ocean', 'image_sha256': '554a69a145a1d12350b0b18801b8729007d9f85827d184a652d49239178a0169', 'reference_mentions': {'dog': 5}}, 394449: {'targets': ['clock'], 'foils': ['bench', 'train'], 'caption': 'a train station with two people sitting on a bench', 'image_sha256': '86d43167fc5417798e0f71a45d0e7c59ceddc6a1582500b556c3d890943ac6bc', 'reference_mentions': {'clock': 3}}, 395831: {'targets': ['bottle'], 'foils': ['donut', 'cup'], 'caption': 'a box of a variety of donuts and a cup of water', 'image_sha256': '6d117dffaeeeeaa3ec349eabfee9652cc85640f4cb6b85357d5985546f03b85d', 'reference_mentions': {'bottle': 2}}, 396932: {'targets': ['tv', 'backpack'], 'foils': ['book'], 'caption': 'a black suitcase sitting on the floor next to a book shelf', 'image_sha256': '7d23fd1801d05fa766238ca6cce8891d400dfa7f4c0c4bddbc27359208529b73', 'reference_mentions': {'tv': 5, 'backpack': 3}}, 399164: {'targets': ['laptop'], 'foils': ['chair'], 'caption': 'a desk with a computer and a chair', 'image_sha256': '94e9f38c675cc13c1430ca54d13384d2361c205a30959a04799e368e0cde2876', 'reference_mentions': {'laptop': 2}}, 406949: {'targets': ['car'], 'foils': ['bus'], 'caption': 'a city street with a bus and a person on a bike', 'image_sha256': '7f36905c2a6b6baa3434b718fd2a2601098ab4fe5e7b18094b1223217b969de9', 'reference_mentions': {'car': 3}}, 411727: {'targets': ['book'], 'foils': ['laptop'], 'caption': 'a laptop computer sitting on top of a desk', 'image_sha256': '061f348e4f57ced469a450c1fcc8dcc5e712a0775ff349556dcae25b811d6b86', 'reference_mentions': {'book': 3}}, 415716: {'targets': ['vase'], 'foils': ['chair'], 'caption': 'a living room filled with furniture and a chair', 'image_sha256': '0d3dd7d85cfcc4cf0ea1d814197ca3c4e419a1ce968b269f5d06fabe0b367f58', 'reference_mentions': {'vase': 2}}, 418692: {'targets': ['chair'], 'foils': ['couch'], 'caption': 'a man sitting on a couch in a living room', 'image_sha256': '5006098d6cba04621c2d232efee2380ed765d00961e9dc84973c0065da7989bf', 'reference_mentions': {'chair': 4}}, 419309: {'targets': ['boat'], 'foils': ['bird'], 'caption': 'a large body of water filled with birds', 'image_sha256': 'a43bf23103e1b4a7b3a83988f78723324c682eb3b725207ea9ce2fedcad2369d', 'reference_mentions': {'boat': 3}}, 421231: {'targets': ['fork'], 'foils': ['cake'], 'caption': 'a man standing in front of a chocolate cake', 'image_sha256': '8afdbd12047ea1824b53516270a661af0a43b4115a1cfaadd31c5bfb34ef85d4', 'reference_mentions': {'fork': 3}}, 423008: {'targets': ['umbrella'], 'foils': ['boat'], 'caption': 'a crowd of people on a body of water with boats', 'image_sha256': '3a91a3879b728e08930eddd97f7f32565e50ab62b22ba917fe73426a1c802a1f', 'reference_mentions': {'umbrella': 3}}, 428975: {'targets': ['book'], 'foils': ['clock'], 'caption': 'a table with a lamp and a clock on it', 'image_sha256': '985622dbbeec7950df94202a4bcf959eca47751b810438ddacaf9fc9a2d95dae', 'reference_mentions': {'book': 4}}, 430525: {'targets': ['handbag'], 'foils': ['bench'], 'caption': 'a group of people that are sitting on a bench', 'image_sha256': 'c011910a75e4780dc9fe22f60c02dab0ad65a844392418b30de9abc68e3ff024', 'reference_mentions': {'handbag': 2}}, 456969: {'targets': ['pizza'], 'foils': ['laptop'], 'caption': 'a woman sitting at a table with a laptop and a plate of food', 'image_sha256': '50a590b0d80116e4a0d671e735cafedc8c5f162a49309cc6b5d08d58baca812a', 'reference_mentions': {'pizza': 2}}, 467386: {'targets': ['bird'], 'foils': ['cat'], 'caption': 'a cat sitting on the ground next to a door', 'image_sha256': '317703cbeaa5297e21415285fe9b512ae404c5d24e6f943154b3e65e12abc041', 'reference_mentions': {'bird': 2}}, 468778: {'targets': ['umbrella'], 'foils': ['teddy bear'], 'caption': 'a group of stuffed teddy bears standing next to a table', 'image_sha256': '8cafb3a5f7dd18566b8f59ec1d284d03e16f581b425a1c1f9ee8f3c6a717288d', 'reference_mentions': {'umbrella': 3}}, 472399: {'targets': ['vase'], 'foils': ['chair', 'cat'], 'caption': 'a white cat sitting in a white chair next to flowers', 'image_sha256': '90d6f09768a91d47ebc40a3d03f11927a8e7067eebc7a63cfe2dc9d9788e8cc5', 'reference_mentions': {'vase': 3}}, 476664: {'targets': ['cat'], 'foils': ['bed'], 'caption': 'a dog laying on a bed in a room', 'image_sha256': 'be02db5a0331cddf23810e123a9a805003c3093e377281eee2b7ab0faa9c703b', 'reference_mentions': {'cat': 5}}, 480076: {'targets': ['chair'], 'foils': ['couch'], 'caption': 'a couch that has a cat sitting on it', 'image_sha256': '4f9f240720f90507032f12fe7f9226d8c3ec050bb8e304ccc6dd1651aefd2ccc', 'reference_mentions': {'chair': 3}}, 493435: {'targets': ['tv'], 'foils': ['bench', 'umbrella'], 'caption': 'a park bench with an umbrella on it', 'image_sha256': '294f8acccef1b16522bb285065b40ad4fa5e270b4cd6905353f15a1addda9bc1', 'reference_mentions': {'tv': 5}}, 497488: {'targets': ['chair'], 'foils': ['bed'], 'caption': 'a bedroom with a bed and a desk', 'image_sha256': '5086109e9b280fe32c723f727f26fb52f3d83408e7a1982d2b26a2618dc1fbaa', 'reference_mentions': {'chair': 2}}, 499571: {'targets': ['bottle'], 'foils': ['cat'], 'caption': 'a black and white cat standing on a table', 'image_sha256': '43bf8e2560a61faabea2f4e7df707bbd2e5492bb7796770a2ef3807563607b84', 'reference_mentions': {'bottle': 2}}, 504902: {'targets': ['dog'], 'foils': ['truck'], 'caption': 'a truck that is parked in the grass', 'image_sha256': '2edbe928a9ed860f7682e4d5c088102f79d3ec902c5ca9160768aa78a456af14', 'reference_mentions': {'dog': 5}}, 519094: {'targets': ['bottle'], 'foils': ['cat'], 'caption': 'a cat that is laying down with its mouth open', 'image_sha256': '6d0597df6b584cd620fd267aeddc0dba4171fbe0ace74aa8a771441c8690aa6d', 'reference_mentions': {'bottle': 2}}, 526622: {'targets': ['boat'], 'foils': ['bicycle'], 'caption': 'a bicycle parked next to a body of water', 'image_sha256': 'f2a2ef5c0da81351b7ac9fd533d8f0223744843d785e46f0a528ac413eb9ad82', 'reference_mentions': {'boat': 4}}, 530384: {'targets': ['chair'], 'foils': ['refrigerator'], 'caption': 'a kitchen with a table and a refrigerator', 'image_sha256': '4c2edcb3cf2d268354d367337e6604eac1d1ae0d84d4f8751ebe71ef7c4b6397', 'reference_mentions': {'chair': 2}}, 535808: {'targets': ['book'], 'foils': ['bicycle'], 'caption': 'a bicycle parked in front of a store', 'image_sha256': '6b327af2d4f7c9ce05bd8a15e1a0ba02306c0a3eb1a2dc10893f060aa519d998', 'reference_mentions': {'book': 3}}, 545903: {'targets': ['bowl'], 'foils': ['bed'], 'caption': 'a bed sitting in a room next to a pair of shoes', 'image_sha256': '47d8951cb908358210f57c8f713efecbd5f407ee8eca0b8b6f6cc4d0728d9031', 'reference_mentions': {'bowl': 2}}, 554980: {'targets': ['fork'], 'foils': ['sandwich'], 'caption': 'a tray with a sandwich an orange and an orange', 'image_sha256': '8e52e97be1cb75307e2d1ba3136ddfcfde7825010b86d9f2cd012db603ae6116', 'reference_mentions': {'fork': 3}}, 560365: {'targets': ['fork'], 'foils': ['pizza'], 'caption': 'a pizza sitting on top of a pan on a table', 'image_sha256': '1002946f5ff5c9b426a0ff449e17eaae5349cdeed68beed9e4cb40f050e9c9b9', 'reference_mentions': {'fork': 2}}, 568041: {'targets': ['vase'], 'foils': ['couch'], 'caption': 'a living room with a couch and a mirror', 'image_sha256': '31b929cef60a07ed9d666325d0b0916f14fb2764313d8e509c7d5bcf23f98eb9', 'reference_mentions': {'vase': 2}}, 571182: {'targets': ['bus'], 'foils': ['umbrella'], 'caption': 'a group of people standing in the snow holding umbrellas', 'image_sha256': '4b0baa4b48dbe0da7ec300bf3fff067674f33b6d9e82b998dda8740f634e5734', 'reference_mentions': {'bus': 5}}, 571544: {'targets': ['cup'], 'foils': ['cat'], 'caption': 'a black cat is sitting on a bed', 'image_sha256': '7a085dcc20d7b2672240f26c0edea704cc9ba711f8886cd6a5176ec9f1073367', 'reference_mentions': {'cup': 4}}, 573105: {'targets': ['suitcase'], 'foils': ['car'], 'caption': 'a red car with three pieces of luggage', 'image_sha256': '4a5efc6d826a02cc2c76fbeed341e11345f78dbaa3e9055361e40d2a00291f97', 'reference_mentions': {'suitcase': 3}}, 579602: {'targets': ['umbrella'], 'foils': ['kite'], 'caption': 'a group of kites flying over a beach', 'image_sha256': 'd2610c6590b32d3df793e4aa2bc590aca968767203dfd8cdb4514b4034c9c76d', 'reference_mentions': {'umbrella': 2}}, 579673: {'targets': ['cup'], 'foils': ['umbrella'], 'caption': 'a table that has a umbrella on it', 'image_sha256': '4be3abedac7e3773524f59577e8b360971147ff6f546aa6e64addc8e152a281d', 'reference_mentions': {'cup': 4}}, 581394: {'targets': ['bicycle'], 'foils': ['bus'], 'caption': 'a green and green bus with a bunch of people on top of it', 'image_sha256': 'd750451ec2fc61f6d9beba6fc05f8d565b493a741ac6c2cbc91d184644a0f180', 'reference_mentions': {'bicycle': 2}}}
VAL_PINS = {'val_0000.json': '7a13112d6b180dea9fdedbf405aca42e64564310977c105f31fe5edced4a93f4', 'val_0001.json': 'a868d753742411bed0554747eae20f0131b9df3d6c158f21372a7252c4928eaf', 'val_0002.json': '4bed69399d25af1eb8bf52e61972ed162188ae2b2c806c5616894952712c86b6', 'val_0003.json': 'afb5f12cf003bd7fa6fda882c0407e0177abc2b848cd597e6a8928aed3022f27', 'val_0004.json': '3a500b41aeb9e471d6f339456a96bc1be1f5a8903bd90d1f8557e34cc404efb0'}
from pathlib import Path
import hashlib
import json
import math
from collections import Counter

ROOT=Path('/kaggle/input/datasets/ducanh2403')
OUTPUT=Path('/kaggle/working/cider_roi_reference_cohort_candidates.json')

def checked(path,pin):
    with path.open('rb') as stream:
        if hashlib.file_digest(stream,'sha256').hexdigest()!=pin:
            raise ValueError(f'Wrong input: {path}')
        stream.seek(0)
        return json.load(stream)

def geometry(a,b):
    area=lambda x:(x[2]-x[0])*(x[3]-x[1])
    aa,bb=area(a),area(b)
    overlap=math.prod(max(0,min(a[i+2],b[i+2])-max(a[i],b[i])) for i in (0,1))
    distance=math.hypot(*[(a[i]+a[i+2]-b[i]-b[i+2])/2 for i in (0,1)])
    return max(aa,bb)/min(aa,bb),distance,overlap/(aa+bb-overlap)

def pair_pass(a,b):
    ratio,distance,iou=geometry(a,b)
    return ratio<=2 and distance<=.35 and iou<=.10

def main():
    meta={}
    for name,pin in VAL_PINS.items():
        path,=(ROOT/'csa-full-roi-karpathy-v134').rglob(name)
        for row in checked(path,pin)['images']:
            if row['coco_id'] in ARCHIVED:
                if row['group']!='val' or row['coco_id'] in meta:
                    raise ValueError('Invalid VALID metadata')
                meta[row['coco_id']]=row
    assert meta.keys()==ARCHIVED.keys()
    links,=(ROOT/'csa-prompt-full-karpathy-v175').rglob('prompt_links.jsonl')
    texts=checked(links.parent/'prompt_texts.json','35f28b5f9f6cbbc2eaa6a0fc3ea816f8722013a5804814da6555edcfe97f04c2')
    digest=hashlib.sha256()
    prompts={}
    with links.open('rb') as f:
        for line in f:
            digest.update(line)
            row=json.loads(line)
            if row['group']=='val' and row['coco_id'] in meta:
                if row['coco_id'] in prompts:
                    raise ValueError('Duplicate VALID prompt row')
                prompts[row['coco_id']]=row
    assert digest.hexdigest()=='e90c79b3b04540630b4bd2d279a4070ef05da7f4b65970c952851c37e2423bf0'
    assert prompts.keys()==meta.keys()
    data=checked(ROOT/'objectdetectionecache/objectdetectioncache.json','498825daf8f8773a1c92f4111dc4ccf1f4296343f952b486e0aa56caf82ec04c')
    data=data.get('data',data)
    by_name={Path(k).name:v for k,v in data.items()}
    assert len(by_name)==len(data)
    candidates=[]
    for cid,row in meta.items():
        saved=prompts[cid]
        entry=ARCHIVED[cid]
        objects=by_name[row['filename']]
        if isinstance(objects,dict):
            objects=objects['objects']
        ordered=sorted(enumerate(objects),key=lambda item:(-float(item[1]['conf']),item[0]))[:150]
        if not ordered or len(saved['objects'])>=8:
            continue
        assert len(ordered)==saved['roi_count']==len(row['region_kinds'])
        scales=(row['width'],row['height'])*2
        rois=[]
        for idx,(_,obj) in enumerate(ordered):
            b=[min(max(float(x),0),s)/s for x,s in zip(obj['bbox'],scales,strict=True)]
            assert b[0]<b[2] and b[1]<b[3]
            label=' '.join(obj['label'].lower().replace('_',' ').split())
            rois.append(dict(roi=idx,label=label,bbox=b,confidence=float(obj['conf'])))
        for obj in saved['objects']:
            r=rois[obj['roi']]
            assert r['label']==obj['label'] and max(abs(a-b) for a,b in zip(r['bbox'],obj['roi_bbox'],strict=True))<=1e-6
        used={o['roi'] for o in saved['objects']}
        existing={o['label'] for o in saved['objects']}
        available=[r for r in rois if r['roi'] not in used and r['confidence']>=.5]
        for a in available:
            target=a['label']
            phrase=('an ' if target[0] in 'aeiou' else 'a ')+target
            if target not in entry['targets'] or target in existing or phrase not in texts:
                continue
            for b in available:
                if b['label'] not in entry['foils'] or not pair_pass(a['bbox'],b['bbox']):
                    continue
                if (target[0] in 'aeiou') != (b['label'][0] in 'aeiou'):
                    continue
                ratio,distance,iou=geometry(a['bbox'],b['bbox'])
                candidates.append(dict(coco_id=cid,filename=row['filename'],image_sha256=entry['image_sha256'],
                    caption=entry['caption'],prompt=phrase,matched_roi=a,wrong_roi=b,area_ratio=ratio,
                    reference_mentions=entry['reference_mentions'][target],
                    center_distance=distance,iou=iou,rank_quality=distance+math.log(ratio)/2))
    count=Counter()
    seen=set()
    shortlist=[]
    for pair in sorted(candidates,key=lambda r:(r['rank_quality'],r['coco_id'],r['matched_roi']['roi'],r['wrong_roi']['roi'])):
        label=pair['matched_roi']['label']
        if pair['coco_id'] in seen or count[label]>=6:
            continue
        seen.add(pair['coco_id'])
        count[label]+=1
        shortlist.append(pair)
        if len(shortlist)==60:
            break
    OUTPUT.write_text(json.dumps(dict(split='val',selection='target in at least two VALID references',screened_images=len(meta),eligible_pairs=len(candidates),
                          shortlisted_images=len(shortlist),rows=shortlist),indent=2,allow_nan=False)+'\n',encoding='utf8')
    print(OUTPUT, len(shortlist), 'candidates before visual review')

if __name__=='__main__':
    main()
```
